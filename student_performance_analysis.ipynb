```python

import pandas as pd
import matplotlib.pyplot as plt

# Student dataset
data = {
    "Name": ["Aman", "Priya", "Rahul", "Neha", "Rohit",
             "Simran", "Vikas", "Anjali"],
    "Attendance": [85, 92, 40, 75, 10, 65, 48, 95],
    "Marks": [90, 95, 35, 80, 20, 70, 45, 88]
}

df = pd.DataFrame(data)

# Performance analysis
df["Performance"] = df["Marks"].apply(
    lambda marks: "Good" if marks >= 60 else "Needs Improvement"
)

print("STUDENT PERFORMANCE ANALYSIS")
print("=" * 45)
print(df.to_string(index=False))

print("\nAverage Marks:", round(df["Marks"].mean(), 2))
print("Average Attendance:", round(df["Attendance"].mean(), 2), "%")
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())

# Attendance and marks graph
df.plot(
    x="Name",
    y=["Attendance", "Marks"],
    kind="bar",
    figsize=(10, 5)
)

plt.title("Student Attendance and Marks Analysis")
plt.ylabel("Score (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
